```python

# ASSIGNMENT 2
# Feedforward Neural Network using Keras and TensorFlow

# a. Import necessary packages
import tensorflow as tf
import matplotlib.pyplot as plt

# b. Load MNIST dataset [note: x== img(28x28), y==digits]
(x_train,y_train),(x_test,y_test) =tf.keras.datasets.mnist.load_data()

# Normalize data
x_train = x_train / 255.0
x_test = x_test / 255.0

# c. Define the neural network
model = tf.keras.Sequential([
    tf.keras.layers.Flatten(input_shape=(28, 28)),
    tf.keras.layers.Dense(128, activation='relu'),
    tf.keras.layers.Dense(10, activation='softmax')
])

# d. Compile model using SGD
model.compile(
    optimizer=tf.keras.optimizers.SGD(learning_rate=0.01),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# Train the model
history = model.fit(x_train, y_train, epochs=5, validation_split=0.1)

# e. Evaluate the network
loss, accuracy = model.evaluate(x_test, y_test)

print("Test Loss:", loss)
print("Test Accuracy:", accuracy)

# f. Plot training loss and accuracy
plt.subplot(1, 2, 1)
plt.plot(history.history['loss'])
plt.title('Training Loss')
plt.xlabel('Epochs')
plt.ylabel('Loss')

plt.subplot(1, 2, 2)
plt.plot(history.history['accuracy'])
plt.title('Training Accuracy')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')

plt.show()
```

```python

# ASSIGNMENT 3
# Image Classification using CNN

# a. Import necessary packages
import tensorflow as tf
import matplotlib.pyplot as plt

# b. Load MNIST dataset
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

# Preprocess the data
x_train = x_train / 255.0
x_test = x_test / 255.0

# Reshape images to include channel dimension
x_train = x_train.reshape(-1, 28, 28, 1)
x_test = x_test.reshape(-1, 28, 28, 1)

# c. Define CNN model
model = tf.keras.Sequential([
    tf.keras.layers.Conv2D(32, (3, 3), activation='relu',
                           input_shape=(28, 28, 1)),
    tf.keras.layers.MaxPooling2D((2, 2)),
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(64, activation='relu'),
    tf.keras.layers.Dense(10, activation='softmax')
])

# Compile the model
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# d. Train the model
history = model.fit(
    x_train, y_train,
    epochs=5,
    validation_split=0.1
)

# e. Evaluate model performance
loss, accuracy = model.evaluate(x_test, y_test)

print("Test Loss:", loss)
print("Test Accuracy:", accuracy)

# Plot training accuracy and loss
plt.subplot(1, 2, 1)
plt.plot(history.history['accuracy'])
plt.title('Training Accuracy')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')

plt.subplot(1, 2, 2)
plt.plot(history.history['loss'])
plt.title('Training Loss')
plt.xlabel('Epochs')
plt.ylabel('Loss')

plt.tight_layout()
plt.show()
```

```python
#ass4

import tensorflow as tf
from tensorflow.keras import layers, models
import numpy as np
import matplotlib.pyplot as plt

# --------------------------------------------------
# 1. Load MNIST dataset
# --------------------------------------------------

(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

print("Training data:", x_train.shape)
print("Testing data:", x_test.shape)


# --------------------------------------------------
# 2. Preprocessing
# --------------------------------------------------

x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

x_train = x_train.reshape(-1, 784)
x_test = x_test.reshape(-1, 784)


# --------------------------------------------------
# 3. Select normal data (digit 0)
# --------------------------------------------------

x_train_normal = x_train[y_train == 0]

print("Normal training samples:", x_train_normal.shape[0])


# --------------------------------------------------
# 4. Define Encoder
# --------------------------------------------------

encoder = models.Sequential([
    layers.Input(shape=(784,)),
    layers.Dense(128, activation="relu"),
    layers.Dense(32, activation="relu")
])


# --------------------------------------------------
# 5. Define Decoder
# --------------------------------------------------

decoder = models.Sequential([
    layers.Dense(128, activation="relu"),
    layers.Dense(784, activation="sigmoid")
])


# --------------------------------------------------
# 6. Define Autoencoder
# --------------------------------------------------

autoencoder = models.Sequential([
    encoder,
    decoder
])

autoencoder.summary()


# --------------------------------------------------
# 7. Compile
# --------------------------------------------------

autoencoder.compile(
    optimizer="adam",
    loss="mse"
)


# --------------------------------------------------
# 8. Train
# --------------------------------------------------

history = autoencoder.fit(
    x_train_normal,
    x_train_normal,
    epochs=10,
    batch_size=128,
    validation_split=0.1,
    shuffle=True
)


# --------------------------------------------------
# 9. Plot loss
# --------------------------------------------------

plt.figure(figsize=(8, 5))

plt.plot(history.history["loss"], label="Training Loss")
plt.plot(history.history["val_loss"], label="Validation Loss")

plt.xlabel("Epoch")
plt.ylabel("MSE Loss")
plt.title("Autoencoder Loss vs Epochs")
plt.legend()
plt.grid()
plt.show()


# --------------------------------------------------
# 10. Reconstruct test images
# --------------------------------------------------

x_test_reconstructed = autoencoder.predict(x_test)


# --------------------------------------------------
# 11. Calculate reconstruction error
# --------------------------------------------------

reconstruction_error = np.mean(
    np.square(x_test - x_test_reconstructed),
    axis=1
)


# --------------------------------------------------
# 12. Calculate threshold using normal data
# --------------------------------------------------

normal_reconstructed = autoencoder.predict(x_train_normal)

normal_error = np.mean(
    np.square(x_train_normal - normal_reconstructed),
    axis=1
)

threshold = np.percentile(normal_error, 95)

print("Anomaly Threshold:", threshold)


# --------------------------------------------------
# 13. Detect anomalies
# --------------------------------------------------

anomalies = reconstruction_error > threshold

print("Total test samples:", len(x_test))
print("Detected anomalies:", np.sum(anomalies))
print("Detected normal samples:", np.sum(~anomalies))


# --------------------------------------------------
# 14. Display images
# --------------------------------------------------

plt.figure(figsize=(12, 6))

for i in range(10):

    plt.subplot(2, 5, i + 1)

    plt.imshow(x_test[i].reshape(28, 28), cmap="gray")

    status = "Anomaly" if anomalies[i] else "Normal"

    plt.title(
        f"{status}\nError: {reconstruction_error[i]:.4f}"
    )

    plt.axis("off")

plt.tight_layout()
plt.show()


# --------------------------------------------------
# 15. Reconstruction error distribution
# --------------------------------------------------

plt.figure(figsize=(10, 5))

plt.hist(reconstruction_error, bins=50)

plt.axvline(
    threshold,
    linestyle="--",
    label="Anomaly Threshold"
)

plt.xlabel("Reconstruction Error")
plt.ylabel("Number of Images")
plt.title("Reconstruction Error Distribution")
plt.legend()

plt.show()
```

```python

# ASSIGNMENT 4
# Anomaly Detection using Autoencoder

# a. Import necessary packages
import tensorflow as tf
import numpy as np

# b. Load MNIST dataset
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

# Normalize and flatten images
x_train = x_train.reshape(-1, 784) / 255.0
x_test = x_test.reshape(-1, 784) / 255.0

# Train only on normal images (digit 0)
normal_train = x_train[y_train == 0]

# c. Define Encoder
encoder = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(784,)),
    tf.keras.layers.Dense(32, activation='relu')
])

# d. Define Decoder
decoder = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(32,)),
    tf.keras.layers.Dense(784, activation='sigmoid')
])

# Combine Encoder and Decoder
autoencoder = tf.keras.Sequential([encoder, decoder])

# e. Compile the model
autoencoder.compile(
    optimizer='adam',
    loss='mse',
    metrics=['mae']
)

# Train the Autoencoder
autoencoder.fit(normal_train, normal_train, epochs=5, batch_size=128)

# Calculate reconstruction error on normal training images
train_reconstructed = autoencoder.predict(normal_train)
train_error = np.mean((normal_train - train_reconstructed) ** 2, axis=1)

# Set anomaly threshold
threshold = np.percentile(train_error, 95)

# Reconstruct test images
test_reconstructed = autoencoder.predict(x_test)
test_error = np.mean((x_test - test_reconstructed) ** 2, axis=1)

# Detect anomalies
predictions = test_error > threshold

# Actual labels: 0 = Normal, 1 = Anomaly
actual = y_test != 0

# Evaluate anomaly detection accuracy
accuracy = np.mean(predictions == actual)

print("Threshold:", threshold)
print("Anomaly Detection Accuracy:", accuracy)

# Display sample predictions
for i in range(10):
    print("Digit:", y_test[i],
          "Predicted:", "Anomaly" if predictions[i] else "Normal")
```

```python
# ass 5



import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Embedding, Lambda
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.utils import to_categorical

# a. Data preparation
corpus = [
    "deep learning models learn hierarchical representations from data",
    "neural networks process information using interconnected layers",
    "word embeddings capture semantic relationships between words"
]
tokenizer = Tokenizer()
tokenizer.fit_on_texts(corpus)
sequences = tokenizer.texts_to_sequences(corpus)
vocab_size = len(tokenizer.word_index) + 1
window_size = 2  # Context words on each side

# b. Generate training data (Context words -> Target center word)
X, y = [], []
for seq in sequences:
    for i, target in enumerate(seq):
        context = [seq[j] for j in range(max(0, i - window_size), min(len(seq), i + window_size + 1)) if j != i]
        if len(context) == window_size * 2:  # Fixed context length for clean batching
            X.append(context)
            y.append(target)

X = np.array(X)
y = to_categorical(y, num_classes=vocab_size)

# c. Train model
embed_dim = 10
model = Sequential([
    Embedding(input_dim=vocab_size, output_dim=embed_dim, input_length=window_size * 2),
    Lambda(lambda x: tf.reduce_mean(x, axis=1)),  # Averages context word embeddings
    Dense(vocab_size, activation='softmax')       # Predicts target center word
])

model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
model.fit(X, y, epochs=100, verbose=0)

# d. Output (Extract learned word embeddings & test prediction)
weights = model.get_weights()[0]
word_to_idx = tokenizer.word_index
idx_to_word = {v: k for k, v in word_to_idx.items()}

print("Embedding vector for 'neural':\n", weights[word_to_idx['neural']])

# Test predicting center word from context
sample_context = np.array([X[0]])
pred_idx = np.argmax(model.predict(sample_context, verbose=0))
print("Context:", [idx_to_word[i] for i in X[0]], "-> Predicted Target:", idx_to_word[pred_idx])
```

```python

# ASSIGNMENT 5
# Continuous Bag of Words (CBOW)

# INSTALLATION (run once before executing)
# pip install gensim nltk
# In Jupyter/Colab: %pip install gensim nltk
# Download NLTK stopwords once:
# import nltk
# nltk.download('stopwords')

# a. Import packages
import re
from gensim.models import Word2Vec
from nltk.corpus import stopwords

# b. Load dataset
with open("alice.txt", "r", encoding="utf-8") as file:
    text = file.read()

# c. Data preprocessing
text = re.sub(r'[^a-zA-Z. ]', ' ', text.lower())
stop_words = set(stopwords.words('english'))

sentences = []
for sentence in text.split('.'):
    words = [w for w in sentence.split()
             if w not in stop_words]
    if words:
        sentences.append(words)

# d. Generate training data
window = 2
training_data = []

for words in sentences:
    for i in range(len(words)):
        context = words[max(0, i-window):i] + words[i+1:i+window+1]
        if context:
            training_data.append((context, words[i]))

print("Sample Training Data:")
print(training_data[:5])

# e. Train CBOW model
model = Word2Vec(
    sentences,
    vector_size=50,
    window=2,
    min_count=1,
    sg=0,
    negative=5,
    epochs=20
)

# f. Find similar words
print("\nSimilar words to Alice:")
print(model.wv.most_similar('alice', topn=5))

# g. Predict output word
context_words = training_data[0][0]

print("\nContext:", context_words)
print("Actual Word:", training_data[0][1])
print("Predicted Words:")
print(model.predict_output_word(context_words, topn=5))
```

```python

# ASSIGNMENT 6
# Transfer Learning using CNN

# Installation (run once if TensorFlow is missing)
# pip install tensorflow
# In Jupyter/Colab: %pip install tensorflow

# a. Import necessary packages
import tensorflow as tf

# b. Load CIFAR-10 dataset
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.cifar10.load_data()

# Use smaller dataset for faster training
x_train, y_train = x_train[:500], y_train[:500]  # Original: 2000
x_test, y_test = x_test[:100], y_test[:100]      # Original: 500

# Preprocess images
x_train = tf.image.resize(x_train, (96, 96))
x_test = tf.image.resize(x_test, (96, 96))

x_train = tf.keras.applications.mobilenet_v2.preprocess_input(x_train)
x_test = tf.keras.applications.mobilenet_v2.preprocess_input(x_test)

# c. Load pretrained CNN model
base_model = tf.keras.applications.MobileNetV2(
    weights='imagenet',
    include_top=False,
    input_shape=(96, 96, 3)
)

# d. Freeze pretrained layers
base_model.trainable = False

# e. Add custom classifier
model = tf.keras.Sequential([
    base_model,
    tf.keras.layers.GlobalAveragePooling2D(),
    tf.keras.layers.Dense(64, activation='relu'),
    tf.keras.layers.Dense(10, activation='softmax')
])

# Compile model
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# f. Train custom classifier
model.fit(x_train, y_train, epochs=1, batch_size=32)  # Original: 3 epochs

# g. Fine-tune model
base_model.trainable = True

# Keep most layers frozen
for layer in base_model.layers[:-20]:
    layer.trainable = False

# Keep BatchNormalization layers frozen
for layer in base_model.layers:
    if isinstance(layer, tf.keras.layers.BatchNormalization):
        layer.trainable = False

# Compile with smaller learning rate
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.00001),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# Train again (fine-tuning)
model.fit(x_train, y_train, epochs=1, batch_size=32)  # Original: 2 epochs

# h. Evaluate model
loss, accuracy = model.evaluate(x_test, y_test)

print("Test Loss:", loss)
print("Test Accuracy:", accuracy)
```
